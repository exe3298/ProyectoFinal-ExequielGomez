## 1. Presentación del Problema
El mercado de autopartes online sufre de una alta tasa de devoluciones y baja conversión debido a la falta de claridad técnica. Como ingeniero, entiendo que un error de compatibilidad (ej. un sensor con distinta ficha o una bomba de agua para otro bloque motor) genera costos logísticos y pérdida de reputación para el vendedor.

**Objetivo:** Crear un asistente que transforme una descripción precaria en un catálogo profesional con validación técnica y soporte visual preciso.

## 2. Metodología: Optimización mediante Few-Shot Prompting
Para este proyecto, se determinó que la técnica de **Few-Shot Prompting** es superior al Zero-Shot.

En ingeniería, la precisión es crítica. Al proporcionar al modelo ejemplos previos de motores (como el EA111 de VW o el Rocam de Ford), establecemos un estándar de respuesta. Esto reduce las "alucinaciones" de la IA y garantiza que el asesoramiento técnico sea consistente, eliminando respuestas genéricas que podrían inducir al error en la compra.

```python
# Módulo de Consultoría Técnica y Generación de Contenido

def consultar_autoprecision_ia(categoria, prompt_text):
    print(f"--- EJECUTANDO MÓDULO: {categoria} ---")
    # Simulación de la respuesta del modelo procesando las instrucciones
    print(f"Instrucción procesada:\n{prompt_text}")
    print("\n[IA]: Resultado generado con éxito siguiendo criterios de ingeniería.")

# 1. Validación Técnica (Few-Shot)
prompt_ingeniero = """
Actúa como un Ingeniero Mecánico. Valida compatibilidad con ejemplos:
Ejemplo 1: VW Gol Trend 1.6 -> Motor EA111, requiere polea Poly-V.
Ejemplo 2: Ford Ka Rocam -> Motor 8v Nafta, correa 121 dientes.

Tarea: Validar Bomba de agua para motor Zetec Rocam 1.6.
Ingeniero:"""

# 2. Redacción de Ventas (Técnica AIDA)
prompt_ventas = "Redactar publicación para Mercado Libre de Bomba de Agua Zetec Rocam 1.6 usando técnica AIDA."

# 3. Generación Visual (One-Shot)
prompt_visual = "Genera un diagrama técnico esquemático del sistema de refrigeración Zetec Rocam 1.6 resaltando la bomba de agua. Estilo: minimalista, fondo blanco."

consultar_autoprecision_ia("INGENIERÍA", prompt_ingeniero)
consultar_autoprecision_ia("VENTAS", prompt_ventas)
consultar_autoprecision_ia("VISUAL", prompt_visual)
```
